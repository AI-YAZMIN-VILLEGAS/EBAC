```python
len_season_min = 7 * 30
len_season_hr = len_season_min/60
print(len_season_hr)
```

```python
total_series = df.loc[df['duration_unit'] == 'season','duration_num'].sum()
print('Total de seasons = ', total_series)
total_movies_hr = total_series * len_season_hr
print('Total de horas en seasons=',total_movies_hr)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

df_movies = df[['title','country','release_year','duration_num','rating']].loc[df['duration_unit']=='min']
sns.lmplot(x='release_year',y='duration_num',data=df_movies,fit_reg=True,scatter_kws={'marker':'D','s':20})
plt.title('release year vs duration (movies)')
plt.xlabel('year of release')
plt.ylabel('duration (min)')
plt.show()
```

```python
df_movies2 = df[['title','country','release_year','duration_num','rating']].loc[(df['duration_unit']=='min')&(df['duration_num']>60)]
sns.lmplot(x='release_year',y='duration_num',data=df_movies2,fit_reg=True,scatter_kws={'marker':'S','s':20})
plt.title('release year vs duration (movies, length > 60 min)')
plt.xlabel('year of release')
plt.ylabel('duration (min)')
plt.show()
```

```python
df_movies2 = df[['title','country','release_year','duration_num','rating']].loc[(df['duration_unit']=='min')&(df['duration_num']>60)]
sns.lmplot(x='release_year',y='duration_num',data=df_movies2,hue='rating',fit_reg=True,scatter_kws={'marker':'S','s':20})
plt.title('release year vs duration (movies, length > 60 min)')
plt.xlabel('year of release')
plt.ylabel('duration (min)')
plt.show()
```

```python
clase = [
('Miguel',37,'MX'),
('Laura',28,'AR'),
('Raul',46,'CL'),
('Roberto',40,'MX'),
('Daniel',33,'AR'),
('Patricio',47,'CL'),
('Lorena',29,'CO'),
('Esteban',26,'BR'),
('Marta',42,'PE'),
('Ignacio',31,'UY'),
('Silvia',38,'MX'),
('Federico',35,'AR'),
('Julia',23,'CL'),
('Guillermo',49,'BR'),
('Beatriz',36,'CO'),
('Nicolas',28,'PE'),
('Teresa',34,'UY'),
('Adrian',44,'MX'),
('Monica',27,'AR'),
('Ruben',39,'CL'),
('Liliana',32,'BR'),
('Cristian',41,'CO'),
('Paola',25,'PE')
]
```

```python
df_clase = pd.DataFrame(clase, columns = ['Nombre','Edad','País'])
```

```python
df_clase['Par-Impar'] = df_clase['Edad'].apply(lambda x:'Par' if int(x)%2 == 0 else 'Impar')
```

```python
df_clase
```

```python
df_clase.groupby('Par-Impar').count()['Nombre']
```

```python
df_clase.groupby('Par-Impar').count()['Nombre'].plot.bar()
```

```python
df_clase['Dcto']=df_clase['Edad'].apply(lambda x: 0.1 if x > 37 else 0)
```

```python
df_clase
```

```python
import pytz
df_clase['ZonaHoraria'] = df_clase['País'].apply(lambda x: pytz.country_timezones[x][0])
```

```python
df_clase
```

```python
from countryinfo import CountryInfo
df_clase['Capital'] = df_clase['País'].apply(lambda x : CountryInfo(x).capital())
```
