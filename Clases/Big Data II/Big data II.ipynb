# Partición de datos

```python
spark = SparkSession.builder.appName('Partitionby() PySpark').getOrCreate()

dataframe = spark.read.option('header',True).csv('vgsales.csv')

dataframe.printSchema()
```

```python
dataframe = dataframe.cache()
dataframe.count()   # Forces Spark to load the data

dataframe.write \
    .option("header", True) \
    .partitionBy("Platform") \
    .mode("overwrite") \
    .csv("partition/platform")
```

```python
dataframe.write \
    .option("header", True) \
    .partitionBy("Year") \
    .mode("overwrite") \
    .csv("partition/year")
```
