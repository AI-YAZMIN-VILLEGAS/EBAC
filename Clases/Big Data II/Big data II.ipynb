# Partición de datos

```python
spark = SparkSession.builder.appName('Partitionby() PySpark').getOrCreate()

dataframe = spark.read.option('header',True).csv('vgsales.csv')

dataframe.printSchema()
```

```python
dataframe = dataframe.cache()
dataframe.count()   # Forces Spark to load the data

dataframe.write \
    .option("header", True) \
    .partitionBy("Platform") \
    .mode("overwrite") \
    .csv("partition/platform")
```

```python
dataframe.write \
    .option("header", True) \
    .partitionBy("Year") \
    .mode("overwrite") \
    .csv("partition/year")
```

```python
dataframe.write \
    .option("header", True) \
    .partitionBy("Genre",'Publisher') \
    .mode("overwrite") \
    .csv("partition/pub_genre")
```

```python
dataframe.write \
    .option("header", True) \
    .partitionBy("Year",'Genre') \
    .mode("overwrite") \
    .csv("partition/year_genre")
```

```python
dataframe.write \
    .option("header", True) \
    .option('maxRecordsPerFile',10) \
    .partitionBy("Genre") \
    .mode("overwrite") \
    .csv("partition/genre2")
```

```python
dataframe.write \
    .option("header", True) \
    .option('maxRecordsPerFile',50) \
    .partitionBy("Genre") \
    .mode("overwrite") \
    .csv("partition/genre3")
```

# Coalesce y repartition

```python
spark = SparkSession.builder.appName('coalesce() PySpark').getOrCreate()

dataco = spark.read.option('header',True).csv('vgsales.csv')

dataco.printSchema()
```

```python
dataco.count()
```

```python
dataco.repartition(20).write.mode('overwrite').option('header',True).csv('partition/rep')
```

```python
dataco.repartition(30).write.mode('overwrite').option('header',True).csv('partition/rep')
```

```python
dataco.repartition(10).write.mode('overwrite').option('header',True).csv('partition/ej10')
```

```python
dataco.show()
```

```python
dataco_2 = dataco.repartition(20)
dataco_2.rdd.getNumPartitions()
```

```python
dataco_3 = dataco_2.coalesce(10)
```

```python
dataco_3.write \
    .mode("overwrite") \
    .option("header", True) \
    .csv("partition/coalesce")
```

```python
dataco_3.rdd.glom().collect()
```

# Uso de parquet

```python
import pyarrow.parquet as pq
import pandas as pd
import pyarrow as pa

df = pd.DataFrame({'lin1':[-20,100,200],
                   'lin2':['este','es un','ejemplo'],
                   'l3':[False,True,True]},
                   index = list('abc'))
```

```python
df
```

```python
import os
import pyarrow as pa
import pyarrow.parquet as pq

# Create the directory if it doesn't exist
os.makedirs("parquet", exist_ok=True)

tab_example = pa.Table.from_pandas(df)
pq.write_table(tab_example, "parquet/ejemplo.parquet")
```

```python
import pyarrow.parquet as pq

table = pq.read_table("parquet/ejemplo.parquet")
print(table.schema)
print(table.to_pandas().head())
```
