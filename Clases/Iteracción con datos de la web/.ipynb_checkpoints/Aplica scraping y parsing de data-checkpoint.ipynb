```python
url = 'https://listado.mercadolibre.com.mx/iphone#D[A:iphone]'
headers = {
    "User-Agent": "Mozilla/5.0"
}

data = requests.get(url, headers=headers)
html = BeautifulSoup(data.text, 'lxml')
apple_list = []
#print(html.prettify()[:2000])
#for item in html.select("a.poly-component__title"):
#    print(item.get_text(strip=True))
    
for item in html.select('h3.poly-component__title-wrapper a'):#('h3.poly-component__title-wrapper a'):#ui-search-result__wrapper ui-search-result__wrapper--large #.ui-search-layout__item
    try:
        print('_____________________')
        title = item.get_text(strip=True)#.select('h2')[0]
        apple_list.append(title)
        print(title)
    except Exception as e:
        #raise e
        print('')
```

```python
import csv
import os

resultFile = open('mercadolibre.csv','w')
for apple in apple_list:
    resultFile.write(apple+"\n")
resultFile.close()
```
