```python
#PAQUETE NUMPY Y PROBLEMAS ESTADISTICOS
import pandas as pd
import numpy as np
import os

df = pd.read_csv('fifa_eda.csv')
```

```python
df.sample(5)
```

```python
age_np = df['Age'].to_numpy()
height_np = df['Height'].to_numpy()
weight_np = df['Weight'].to_numpy()
overall_np = df['Overall'].to_numpy()
potential_np = df['Potential'].to_numpy()
print(age_np)
print(height_np)
print(weight_np)
print(overall_np)
print(potential_np)
```

```python
media = age_np.mean()
format_medida = '{:.2f}'.format(media)
print('La media de las edades es: ',format_medida,' años')
```

```python
for i in range(0,100,5):
    print(i,"%: ",np.percentile(age_np, i))
```

```python
vals, counts = np.unique(age_np,return_counts=True)
index = np.argmax(counts)
moda = vals[index]
print(f'La moda de la edad de los jugadores es: = {moda:.2f} años')
```

```python
counts
```

```python
mediana = np.median(age_np)
print(f'La mediana de la edad de los jugadores es: = {mediana:.2f} años')
```

```python
desvstd = age_np.std()
print(f'La desviación estándar de la edad de los jugadores es: = {desvstd:.2f} años')
```
