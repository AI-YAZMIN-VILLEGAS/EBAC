
# Simulación Monte Carlo para Presupuesto de Costos

Este notebook incluye:
- Modelo determinístico de costos
- Definición de distribuciones de probabilidad
- Simulación Monte Carlo (5000 iteraciones)
- Interpretación de resultados
- Visualización
- Análisis de sensibilidad
- Conclusiones


## 1. Modelo determinístico

```python

# Costos fijos (determinísticos)
materiales = 5000
mano_obra = 3000
transporte = 1000

# Presupuesto total
costo_total = materiales + mano_obra + transporte

print(f"Costo total determinístico: ${costo_total}")
```

## 2. Definición de distribuciones

```python

import numpy as np

# Definir distribuciones
materiales_dist = np.random.normal(loc=5000, scale=500, size=5000)
mano_obra_dist = np.random.normal(loc=3000, scale=400, size=5000)
transporte_dist = np.random.normal(loc=1000, scale=200, size=5000)
```

## 3. Simulación Monte Carlo

```python

# Simulación
costo_total_sim = materiales_dist + mano_obra_dist + transporte_dist

# Estadísticas
media = np.mean(costo_total_sim)
std = np.std(costo_total_sim)

print(f"Media: ${media:.2f}")
print(f"Desviación estándar: ${std:.2f}")
```

## 4. Visualización

```python

import matplotlib.pyplot as plt

plt.hist(costo_total_sim, bins=50)
plt.title("Distribución del costo total")
plt.xlabel("Costo")
plt.ylabel("Frecuencia")
plt.show()
```


## 5. Interpretación

- La media representa el costo esperado.
- La desviación estándar muestra la variabilidad.
- La distribución permite evaluar riesgos (costos altos).


## 6. Análisis de sensibilidad

```python

# Distribución alternativa (Uniforme)
materiales_uni = np.random.uniform(4500, 5500, 5000)
mano_obra_uni = np.random.uniform(2500, 3500, 5000)
transporte_uni = np.random.uniform(800, 1200, 5000)

costo_total_uni = materiales_uni + mano_obra_uni + transporte_uni

# Comparación
print("Media Normal:", np.mean(costo_total_sim))
print("Media Uniforme:", np.mean(costo_total_uni))
```


## 7. Conclusiones finales

- La simulación Monte Carlo permite modelar incertidumbre.
- Diferentes distribuciones impactan los resultados.
- El análisis de sensibilidad ayuda a entender riesgos.
- Es útil para toma de decisiones bajo incertidumbre.
