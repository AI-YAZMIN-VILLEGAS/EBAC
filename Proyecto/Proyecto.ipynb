### LIMPIAR DATOS
Se cargan los 5 archivos en un dataframe y se hace una limpieza para cada dataframe 

### df_category

```python
#se carga el archivo DIM_CATEGORY, se normalizan las columnas y se limpian los datos de "\r\n".
df_category = pd.read_csv("DIM_CATEGORY.csv")
df_category['ID_CATEGORY'] = pd.to_numeric(df_category['ID_CATEGORY'])
df_category['CATEGORY'] = df_category['CATEGORY'].str.replace(r"[\r\n]+", " ", regex=True) 
df_category['CATEGORY'] = df_category['CATEGORY'].astype("category")
df_category.info()
df_category.head()
```

```python
#se checa si existe valores nulos
df_category.isnull().sum()
```

```python
# se checa si hay renglones duplicados 
df_category.duplicated().sum()
df_category['CATEGORY'].duplicated().sum()
```

### UNIÓN DE DATAFRAME
se van a unir las tablas una por una usando los ID o columnas en común.

```python
df_product.head()
```

```python
df_fact_sales.head()
```

```python
#se hace la union con df_fact_sales y df_product
df_join = df_fact_sales.merge(
    df_product,
    left_on="ITEM_CODE",
    right_on="ITEM",
    how="inner"
)
```

```python
df_join.head()
```

```python
df_category.head()
```

```python
# se hace la unión del resultado de la union anterior con df_category
df_join = df_join.merge(
    df_category,
    left_on="CATEGORY",
    right_on="ID_CATEGORY",
    how="inner"
)
```

```python
df_join.head()
```

```python
#Se hace la unión del resultado de la unión anterior y el dataframe df_segment
df_join = df_join.merge(
    df_segment,
    left_on=["CATEGORY_x", "ATTR1", "ATTR2", "ATTR3", "FORMAT"],
    right_on=["CATEGORY", "ATTR1", "ATTR2", "ATTR3", "FORMAT"],
    how="inner"
)
```

```python
df_join.head()
```

```python
df_calendar.head()
```

```python
#se hace la union del resultado de la unón anterior con df_calendar
df_join = df_join.merge(
    df_calendar,
    on="WEEK",
    how="inner"
)
```

```python
df_join.head()
```

```python
#las columnas del dataframe df_join
df_join.columns
```

```python
#se crean nuevas columnas como la columna del año, mes y año, nombre del mes y precio unitario promedio. Se normaliza el texto del ITEM_DESCRIPTION
import numpy as np
df_join["YEAR"] = df_join['DATE'].dt.year
df_join["YEAR_MONTH"] = df_join["DATE"].dt.to_period("M")
df_join["MONTH_NAME"] = df_join["DATE"].dt.strftime("%b")
df_join["AVG_PRICE_PER_UNIT"] = np.where(
    df_join["TOTAL_UNIT_SALES"] == 0,
    0,
    df_join["TOTAL_VALUE_SALES"] / df_join["TOTAL_UNIT_SALES"]
)
df_join["ITEM_DESCRIPTION"] = df_join["ITEM_DESCRIPTION"].str.upper().str.strip()

df_final=df_join[['DATE','YEAR','YEAR_MONTH','MONTH_NAME','ITEM_CODE','ITEM_DESCRIPTION','CATEGORY','CATEGORY_y','SEGMENT','TOTAL_UNIT_SALES','TOTAL_VALUE_SALES','TOTAL_UNIT_AVG_WEEKLY_SALES','AVG_PRICE_PER_UNIT','REGION']]

```

```python
#se cambia el nombre de la columna CATEGORY_y por CATEGORY DESCRIPTION
df_final = df_final.rename(columns={
    "CATEGORY_y": "CATEGORY_DESCRIPTION",
})
```

```python
df_final.head()
```

```python
#se checa cuantos renglones y columnas hay en el dataframe
df_final.shape
```

```python
df_final.head()
```

```python
#se guarda el dataframe en un archivo csv
df_final.to_csv("proyecto.csv", encoding="utf-8-sig", index=False)
```

```python
#se agrupan las ventas por YEAR_MONT, CATEGORY y REGION
ventas = df_final.groupby(
    ["YEAR_MONTH", "CATEGORY", "REGION"]
).agg({
    "TOTAL_VALUE_SALES": "sum",
    "TOTAL_UNIT_SALES": "sum"
}).reset_index()
```

```python
# se agrupan las ventas mensuales
ventas_mensuales = df_join.groupby("YEAR_MONTH").agg({
    "TOTAL_VALUE_SALES": "sum",
    "TOTAL_UNIT_SALES": "sum"
}).reset_index()

ventas_mensuales.head()
```

```python
#se agrupan las ventas mensuales por categoría
ventas_category_year_month = df_join.groupby(
    ["CATEGORY", "YEAR_MONTH"]
).agg({
    "TOTAL_VALUE_SALES": "sum"
}).reset_index()

ventas_category_year_month.head()
```

```python
#se agrupan las ventas por segménto
ventas_segment = df_join.groupby("SEGMENT").agg({
    "TOTAL_VALUE_SALES": "sum"
}).reset_index()

ventas_segment.head()
```

```python
#se agrupan las ventas por region
ventas_region = df_join.groupby("REGION").agg({
    "TOTAL_VALUE_SALES": "sum"
}).reset_index()

ventas_region.head()
```
