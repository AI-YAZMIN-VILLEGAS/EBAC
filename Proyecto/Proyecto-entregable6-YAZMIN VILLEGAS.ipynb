# Dividir los datos en conjuntos de entrenamiento y prueba:

# Modelo de series de tiempo ARIMA

### AR(1)

```python
from statsmodels.tsa.arima.model import ARIMA

serie = train
model = ARIMA(serie, order=(1, 0, 0))  # AR(1)
result = model.fit()

print(result.summary())
```

```python
print('miu = {} ,phi = {}'.format(result.params[0],result.params[1]))
print(result.summary())
```

# Predicción de modelos

### Predicción de modelos (trainning)

```python
import matplotlib.pyplot as plt

# Predicciones
pred = result.predict(start=0, end=len(train)-1)

# Gráfica
plt.figure(figsize=(22,6))
plt.plot(train, label='Serie de training')
plt.plot(pred, color='red', label='Predicción')
plt.legend()
plt.title('Modelo Autorregresivo ARIMA')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Predicción de modelos de testing

```python
import matplotlib.pyplot as plt
import pandas as pd

# Predicciones
pred2 = result.predict(start=0, end=len(test)-1)

# Give pred the same index as test
pred2.index = test.index

# Gráfica
plt.figure(figsize=(22,6))
plt.plot(test, label='Serie de testing')
plt.plot(pred2, color='red', label='Predicción')
plt.legend()
plt.title('Modelo Autorregresivo ARIMA')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Seleccionnar los parámetros óptimos mediante métodos como AIC

```python
import warnings
import pandas as pd
from statsmodels.tsa.arima.model import ARIMA

warnings.filterwarnings("ignore")

best_aic = float("inf")
best_order = None
best_model = None

for p in range(4):
    for d in range(3):
        for q in range(4):
            try:
                model = ARIMA(train, order=(p, d, q))
                result = model.fit()

                if result.aic < best_aic:
                    best_aic = result.aic
                    best_order = (p, d, q)
                    best_model = result

            except:
                continue

print("Mejor orden:", best_order)
print("AIC:", best_aic)
```

### El mejor modelo es aquel que presenta el menor valor del criterio de información de Akaike (AIC). En este caso, el modelo seleccionado es **ARIMA(1,1,3)**, ya que obtuvo el valor de AIC más bajo entre los modelos evaluados.


# Gráfica de predicción sobre el conjunto de prueba del mejor modelo AR(1,1,3).

```python
result = best_model

pred = result.forecast(steps=len(test))
pred.index = test.index
```

```python
# Gráfica
plt.figure(figsize=(22,6))
plt.plot(test, label='Serie de testing')
plt.plot(pred, color='red', label='Predicción')
plt.legend()
plt.title('Modelo Autorregresivo ARIMA')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Gráfico del modelo optimizado (AR(1,1,3)) para predecir las ventas futuras para los próximos meses

```python
import matplotlib.pyplot as plt
import pandas as pd

# Predicciones

pred = result.forecast(steps=len(future_dates))

# Give pred the same index as test

pred.index = future_dates

combined = pd.concat([serie.iloc[-1:], pred.iloc[1:]])

# Gráfica
plt.figure(figsize=(22,6))
plt.plot(serie, label='Serie')
plt.plot(combined, color='red', label='Predicciones futuras')
plt.legend()
plt.title('Modelo Autorregresivo ARIMA')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

La línea azul corresponde a las ventas históricas del producto. Se observa una variabilidad considerable, con picos superiores a 500 unidades y mínimos cercanos a 240 unidades, lo que indica que las ventas presentan fluctuaciones importantes a lo largo del tiempo.

La línea roja representa las predicciones futuras del modelo ARIMA. Al inicio del pronóstico aún existe cierta variabilidad, ya que el modelo incorpora la información más reciente de la serie. Sin embargo, conforme aumenta el horizonte de predicción, las estimaciones convergen gradualmente hacia un valor cercano a 333 unidades.

El modelo ARIMA(1,1,3) pronostica que las ventas tenderán a estabilizarse alrededor de 333 unidades durante los próximos meses. Aunque la serie histórica presenta una alta variabilidad, el modelo no detecta una tendencia sostenida ni un patrón estacional, por lo que las predicciones convergen hacia un valor estable conforme aumenta el horizonte de pronóstico. Este comportamiento es característico de los modelos ARIMA cuando no existe evidencia suficiente para anticipar cambios sistemáticos en el futuro.


# Conclusiones

Se seleccionó el modelo **ARIMA(1,1,3)** debido a que presentó el menor valor del **criterio de información de Akaike (AIC)** entre los modelos evaluados, lo que indica un mejor equilibrio entre el ajuste y la complejidad del modelo.

El desempeño del modelo se evaluó utilizando el conjunto de prueba, obteniendo un **RMSE de 140.94**, un **MAE de 116.96** y un **MAPE de 24.27%**. Estas métricas indican que, en promedio, las predicciones presentan un error porcentual absoluto del **24.27%** con respecto a los valores reales.

Finalmente, el modelo pronostica que las ventas del producto tenderán a estabilizarse alrededor de **333 unidades** durante los próximos meses. Este comportamiento sugiere que, con la información histórica disponible, el modelo no identifica una tendencia o estacionalidad significativa en el horizonte de predicción, por lo que las estimaciones convergen hacia un nivel estable de ventas.
