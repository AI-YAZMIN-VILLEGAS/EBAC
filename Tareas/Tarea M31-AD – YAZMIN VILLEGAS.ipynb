Generar una lambda que añada la zona horaria y la capital del país de nacionalidad

```python
import pytz
import pycountry

# se hace un diccionario de los países que no estan el la librería pycountry
correcciones = {
    'England': 'United Kingdom',
    'Scotland': 'United Kingdom',
    'Wales': 'United Kingdom',
    'Northern Ireland': 'United Kingdom'
}

#Se hace una función para obtener el código del país para así obetner luego usando lambda y apply la zona horaria.
def get_countrycode(country_name):
    try:
        country_name = correcciones.get(country_name, country_name)
        country_code = pycountry.countries.lookup(country_name).alpha_2
        return country_code
    except:
        return None
#se crea una columna para el codigo del país 
df['Country_code'] = df['Nationality'].apply(get_countrycode)
# se crea la columa de la zona horaria
df['Zona_horaria'] = df['Country_code'].apply( lambda x: pytz.country_timezones.get(x, [None])[0] if x else None)
```

```python
#se importa countryinfo de la librería CountryInfo
from countryinfo import CountryInfo
#se crea un diccionario de corecciones de los países que no estan en la librería CountryInfo
correcciones = {
    'England': 'United Kingdom',
    'Scotland': 'United Kingdom',
    'Wales': 'United Kingdom',
    'Northern Ireland': 'United Kingdom',
    'China PR': 'China',
    'Korea Republic': 'South Korea',
    'Korea DPR': 'North Korea',
    'St Kitts Nevis': 'Saint Kitts and Nevis'
}

#se crea la función safe_capital para obtener la capital y en caso de no tenerla la librería regresa None
def safe_capital(country):
    try:
        return CountryInfo(country).capital()
    except:
        return None

#Se crea un diccionario de capitales para luego usar una funcion (get_capital) que obtenga la capital del país. Se crea el diccionario de capitales para
#no tener que llamar la funcion CountryInfo(country).capital() para cada fila del dataframe y así ahorrar tiempo. 

capitales = {c: safe_capital(correcciones.get(c, c)) for c in df['Nationality'].unique()}

def get_capital(pais):
    return capitales.get(pais)
#se crea la columna Capital y se usa apply y lambda para obtener la capital usando la función get_capital
df['Capital'] = df['Nationality'].apply(lambda x : get_capital(x))
```

Generar una lambda que en base al potencial, añada una columna “Candidato” o “Next Year”, para el premio para el balón de oro.  Jugadores con más de 85, son candidatos.
Se crea la columna Candidato_balon_de_oro y se usa apply y lambda para saver que si es candidato el que tiene Potential mayor que 85 o es Next Year en caso contrario.

```python
df['Candidato_balon_de_oro'] = df['Potential'].apply(lambda x : 'Candidato' if x > 85 else 'Next Year')
```
