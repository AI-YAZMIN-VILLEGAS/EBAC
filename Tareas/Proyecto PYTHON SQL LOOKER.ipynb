# PROYECTO: Prototipo de Delitos

Se seleccionó un archivo con los datos de enero, febrero y marzo de la página oficial, en formato CSV, para posteriormente crear una base de datos en SQL.


```python
#Se crea una función para quitar los acentos de la base de datos, ya que MySQL Workbench solo acepta archivos en formato UTF-8.

import pandas as pd
import unicodedata

def quitar_acentos(texto):
    if isinstance(texto, str):
        texto = unicodedata.normalize('NFKD', texto)
        return ''.join(c for c in texto if not unicodedata.combining(c))
    return texto

#leer el archivo csv
df = pd.read_csv('delitos.csv')

#las columnas del dataframe
columnas = df.columns.tolist()

#quitar los acentos a cada columna y a los nombres de las columnas
for col in columnas:
    df[col] = df[col].apply(quitar_acentos)

df.columns = [quitar_acentos(col) for col in df.columns]

columns = ['Ano','Clave_Ent','Entidad','Bien juridico afectado', 'Tipo de delito','Subtipo de delito','Modalidad','Enero','Febrero','Marzo']

df = df[columns]

df.sample(3)

#df.to_csv('output.csv')
```

```python
# columnas que NO se transforman

id_vars = ['Ano','Clave_Ent','Entidad','Bien juridico afectado', 'Tipo de delito','Subtipo de delito','Modalidad']

df_largo = df.melt(
    id_vars=id_vars,
    value_vars=['Enero', 'Febrero', 'Marzo'],
    var_name='Mes',
    value_name='Total_delitos'
)

df_largo.sample(3)
```

```python
df_largo['Ano_mes'] = df_largo['Ano'].astype(str) + '-' + df_largo['Mes']
df_largo.sample(3)
df_largo.to_csv('output.csv')
```

## Conexión a la base de datos

```python
import mysql.connector

# Crear conexión
conexion = mysql.connector.connect(
    host="localhost",      
    user="root",
    password="Root123$",
    database="PROYECTO_DELITOS"
)

# Crear cursor
cursor = conexion.cursor()

# Ejecutar una consulta
query = "SELECT * FROM delitos LIMIT 10;"
cursor.execute(query)


# Obtener resultados
resultados = cursor.fetchall()

# Mostrar resultados
for fila in resultados:
    print(fila)
```

```python
#(a)  Número promedio de delitos agrupados por municipio y año-mes de ocurrencia de estos. 
# Ejecutar una consulta
query1 = """
SELECT Entidad, Ano_mes, AVG(Total_delitos) as Promedio_delitos
FROM delitos
GROUP BY Entidad, Ano_mes
ORDER BY Entidad, Ano_mes
"""
cursor.execute(query1)


# Obtener resultados
resultados = cursor.fetchall()

# Mostrar resultados
for fila in resultados:
    print(fila)
```

```python
#(b)   El delito más común por municipio
# Ejecutar una consulta
query2 = """
SELECT 
    Entidad,
    `Tipo de delito`,
    Total_Delitos
FROM (
    SELECT 
        Entidad,
        `Tipo de delito`,
        SUM(Total_delitos) AS Total_Delitos,
        RANK() OVER (
            PARTITION BY Entidad
            ORDER BY SUM(Total_delitos) DESC
        ) AS ranking
    FROM delitos
    GROUP BY Entidad, `Tipo de delito`
) AS subconsulta
WHERE ranking = 1;
"""
cursor.execute(query2)


# Obtener resultados
resultados = cursor.fetchall()

# Mostrar resultados
for fila in resultados:
    print(fila)
```

```python
#(c)   La temporada con más crímenes en México
# Ejecutar una consulta
query3 = """
SELECT 
    Ano_mes,
    SUM(Total_delitos) AS Total_Delitos
FROM delitos
GROUP BY Ano_mes
ORDER BY Ano_mes DESC
LIMIT 1;
"""
cursor.execute(query3)


# Obtener resultados
resultados = cursor.fetchall()

# Mostrar resultados
for fila in resultados:
    print(fila)
```

```python
#(d)  El número promedio de delitos agrupados por tipo de delito y año y mes en la ciudad de México ordenados de mayor a menor
# Ejecutar una consulta
query3 = """
SELECT 
    `Tipo de delito`,
    Ano_mes,
    AVG(Total_delitos) AS Promedio_delitos
FROM delitos
GROUP BY `Tipo de delito`, Ano_mes
ORDER BY AVG(Total_delitos) DESC

"""
cursor.execute(query3)


# Obtener resultados
resultados = cursor.fetchall()

# Mostrar resultados
for fila in resultados:
    print(fila)
```

```python
#(d)  El número promedio de delitos agrupados por tipo de delito y año y mes en la ciudad de México ordenados de mayor a menor
# Ejecutar una consulta
query4 = """
SELECT 
    `Tipo de delito`,
    SUM(Total_delitos) AS Tota_Delitos
FROM delitos
GROUP BY `Tipo de delito`
ORDER BY SUM(Total_delitos) DESC

"""
cursor.execute(query4)


# Obtener resultados
resultados = cursor.fetchall()

# Mostrar resultados
for fila in resultados:
    print(fila)
```
