# Aplicación de modelo de regresión logística a la variable Attrition

### Data Split

```python
from sklearn.model_selection import train_test_split

X = df2.drop('Attrition',axis=1)
y = df2['Attrition']

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.30, random_state=1)
```

### Regresión logística 
### Entrenamiento, evaluación y predicción de los resultados del modelo

```python
from sklearn import metrics
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    solver='liblinear',
    class_weight='balanced',
    random_state=42
)

model.fit(X_train, y_train)
y_predict = model.predict(X_test)
```

### y_predict usando el modelo

```python
y_predict[:200]
```

### la predicción de la variable de respuesta y la variable de respuesta real


```python
z = X_test.copy()
z['Attrition real'] = y_test
z['Attrition predicción'] = y_predict
z[['Attrition real','Attrition predicción']].head(20)
```

# Análisis de Sensibilidad e importancia de variables.

```python
coeficientes = pd.DataFrame({
    'Variable': X_train.columns,
    'Coeficiente': model.coef_[0]
})

coeficientes['Importancia'] = abs(coeficientes['Coeficiente'])
coeficientes.sort_values('Importancia', ascending=False)
```

Tus variables categóricas fueron codificadas con cat.codes, por lo que coeficientes como MaritalStatus, Department, EducationField o JobRole deben interpretarse con cuidado. El modelo está asumiendo un orden numérico entre categorías que en realidad no existe.
La importancia de las variables se evaluó mediante la magnitud absoluta de los coeficientes del modelo de regresión logística. Las variables con mayor influencia en las predicciones fueron MaritalStatus (0.633), Department (0.284), YearsWithCurrManager (0.145), TrainingTimesLastYear (0.127), NumCompaniesWorked (0.117) y YearsSinceLastPromotion (0.110).
El análisis de sensibilidad indica que pequeños cambios en las variables con coeficientes de mayor magnitud generan mayores variaciones en la probabilidad estimada por el modelo. En particular, MaritalStatus fue la variable con mayor influencia, seguida de Department y YearsWithCurrManager. Esto sugiere que estas características desempeñan un papel relevante en la clasificación de empleados con riesgo de abandono.

```python
import numpy as np

odds_ratio = pd.DataFrame({
    'Variable': X_train.columns,
    'Odds_Ratio': np.exp(model.coef_[0])
})

odds_ratio.sort_values('Odds_Ratio', ascending=False)
```

MaritalStatus (OR = 1.88) es la variable con mayor efecto positivo. Un incremento de una unidad en esta variable codificada multiplica las probabilidades de rotación por aproximadamente 1.88 veces.
NumCompaniesWorked (OR = 1.12) indica que empleados que han trabajado en más empresas presentan una mayor probabilidad de abandonar la organización.
YearsSinceLastPromotion (OR = 1.12) sugiere que cuanto más tiempo transcurre sin una promoción, mayor es la probabilidad de rotación.

El análisis de importancia de variables mediante Odds Ratios muestra que MaritalStatus, NumCompaniesWorked y YearsSinceLastPromotion son los factores con mayor asociación positiva con la variable de respuesta. Por otro lado, Department, YearsWithCurrManager, TrainingTimesLastYear y MonthlyIncome presentan una asociación negativa con la variable de respuesta.

```python
from sklearn.inspection import permutation_importance

result = permutation_importance(
    model,
    X_test,
    y_test,
    n_repeats=10,
    random_state=42
)

importance_df = pd.DataFrame({
    'Variable': X_test.columns,
    'Importancia': result.importances_mean
})

importance_df.sort_values('Importancia', ascending=False)
```

El análisis de importancia de variables mediante Permutation Importance mostró que MaritalStatus, YearsSinceLastPromotion, NumCompaniesWorked, YearsWithCurrManager, Age y TotalWorkingYears fueron las variables con mayor contribución al desempeño del modelo de regresión logística. Por otro lado, variables como StockOptionLevel, EducationField, PercentSalaryHike y DistanceFromHome presentaron una importancia nula o negativa, lo que sugiere una contribución limitada a la capacidad predictiva del modelo.

```python
importance_df.sort_values(by="Importancia").plot.barh(
    x="Variable",
    y="Importancia",
    legend=False
)
```

### Medidas del modelo

```python
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score, roc_auc_score, classification_report,confusion_matrix

y_pred = model.predict(X_test)
model_score = accuracy_score(y_test, y_pred)
recall_score = recall_score(y_test,y_pred)
precision_score = precision_score(y_test,y_pred)
f1_score = f1_score(y_test,y_pred)
roc_auc_score = roc_auc_score(y_test,y_pred)

print('Medidas del modelo')
print('----------------------------')
print('Accuracy = {0:0.2f}'.format(model_score))
print('Precision = {0:0.2f}'.format(precision_score))
print('Recall = {0:0.2f}'.format(recall_score))
print('F1 Score = {0:0.2f}'.format(f1_score))
print('Roc Auc Score = {0:0.2f}'.format(roc_auc_score))
```

Aunque la accuracy (84%) parece buena, las métricas más importantes (Recall = 0.06, F1 = 0.12, ROC AUC = 0.53) indican que el modelo es muy sensible para encontrar positivos, pero clasifica como positivos muchos casos que en realidad son negativos.
