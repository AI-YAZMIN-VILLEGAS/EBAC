```python
#Leer archivo fifa_eda.csv  
import pandas as pd

df = pd.read_csv('fifa_eda.csv')
df.head()
```

```python
#Elemento 1: Obtener matriz de correlación y dejarla como heat map 
import plotly.express as px
import numpy

corr = df.corr(numeric_only=True)

fig = px.imshow(
    corr,
    text_auto=True,
    color_continuous_scale='RdBu',
    title='Matriz de Correlación',
    width=900,
    height=900
)

fig.show()
```

```python
#Elemento 2: Gráfico de Correlación que responda – “Cuál es la relación entre la edad y el overall?” – explicar la salida 
#El coeficiente de correlación es 0.45 lo que quiere decir que es una correlación positivamente moderada. 
#La variable de la edad y overall tienden aumentar juntas pero no es una relación fuerte.
fig = px.imshow(
    df[['Age','Overall']].corr(),
    color_continuous_scale='RdBu',
    text_auto=True,
    title='Matriz de correlación entre Age y Overall'
)

fig.show()
```

```python
fig = px.scatter(df,x='Age',y='Overall', trendline='ols', title='Relación entre Edad y Overall')
fig.show()

# El gráfico de dispersión muestra que existe una relación positiva moderada
# entre las variables Age y Overall. Esto significa que, en general, cuando
# la edad de los jugadores aumenta, el valor de Overall también tiende a
# incrementarse. Sin embargo, la dispersión de los puntos indica que el
# Overall no depende únicamente de la edad, sino también de otros factores.
```

```python
#Elemento 3: Generar un gráfico de barras por club que indique el número de jugadores 
num_jugadores = df.groupby('Club').size().rename('Número de jugadores').reset_index()

fig = px.bar(num_jugadores,x='Club',y='Número de jugadores',title='Número de jugadores')
fig.update_layout(
    xaxis_title='Club',
    yaxis_title='Número de jugadores',
)
fig.show()
```

```python
#Elemento 3: Generar un gráfico de barras por club que indique el número de jugadores 
#Top 10 clubes con más jugadores

num_jugadores = df.groupby('Club').size().rename('Número de jugadores').reset_index().sort_values('Número de jugadores',ascending=False)[:10]

fig = px.bar(num_jugadores,x='Club',y='Número de jugadores',title='Número de jugadores')
fig.update_layout(
    xaxis_title='Club',
    yaxis_title='Número de jugadores',
    title='Top 10 clubes con más jugadores'
)
fig.show()
```

```python
#Elemento 4: Hacer un gráfico multipanel que indique la relación altura (height) vs skill moves, 
#siendo la variable del panel si es zurdo o derecho 
fig = px.scatter(df,x='Height',y='Skill Moves',facet_col='Preferred Foot',title='Relación de Height y Skill Moves')
fig.show()
```

```python
#Elemento 5: Un gráfico que crea importante para mostrar a los potenciales cracks
potenciales_cracks = df[['Name','Potential']].sort_values('Potential',ascending=False).head(10)
fig = px.bar(potenciales_cracks,x='Name',y='Potential',title='Top 10 de potenciales cracks')
fig.show()
```
