Función lambda para obtener la raíz cuadrada de un número.
Función map, para obtener el largo de una cadena de palabras. 
Tip! La cadena tiene que ser dividida en palabras antes de empezar. 
Función reduce, que sirva para calcular el producto de una lista 
Función filter que sirva para encontrar palabras que contengan mayúsculas o números en un listado 

```python
#obtener la raíz cuadrada de un número
def raiz_cuadrada(num):
    return num**0.5
print(raiz_cuadrada(25))
```

```python
#Función lambda para obtener la raíz cuadrada de un número.
raiz_cuadrada = lambda x:x**0.5
print(raiz_cuadrada(25))
```

```python
#obtener el largo de una cadena de palabras. 
cadena = "Hola mundo python es un lenguaje de programación"
def largo_cadena(cadena):
    lista_largos = []
    for palabra in cadena.split():
        lista_largos.append(len(palabra))
    return lista_largos
print(largo_cadena(cadena))
```

```python
#Función map, para obtener el largo de una cadena de palabras. 
cadena = "Hola mundo python es un lenguaje de programación"
largo_de_una_cadena = list(map(len,cadena.split()))
print(largo_de_una_cadena)
```

```python
#calcular el producto de una lista 
numeros = [1,1,1,2,3]
def calcular_producto(numeros):
    producto = 1
    for n in numeros:
        producto *= n
    return producto
print(calcular_producto(numeros))
```

```python
#Función reduce, que sirva para calcular el producto de una lista 
from functools import reduce
lista = [1,1,1,2,3]
producto = reduce(lambda x,y:x*y,lista)
print(producto)
```

```python
#encontrar palabras que contengan mayúsculas o números en un listado 
listado = ['Hola','hola','hola123','345','EBAC']
def encontrar_mayusculas_numeros(listado):
    resultado = []
    for palabra in listado:
        for letter in palabra:
            if letter.isdigit() or letter.isupper():
                resultado.append(palabra)
                break
    return resultado
print(encontrar_mayusculas_numeros(listado))
            
```

```python
#Función filter que sirva para encontrar palabras que contengan mayúsculas o números en un listado 
listado = ['Hola','hola','hola123','345','EBAC']
resultado =list(filter(lambda palabra:any(letra.isdigit() or letra.isupper() for letra in palabra), listado))
print(resultado)
```
