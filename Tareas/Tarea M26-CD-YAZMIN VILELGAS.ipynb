```python
#Height of Male and Female by Country 2022
#Leer el archivo CSV mencionado
import pandas as pd
df = pd.read_csv('Height of Male and Female by Country 2022.csv')
```

```python
#Imprimir las primeras 50 líneas
df[:50]
```

```python
#Calcular la media, mínima y máxima altura de hombres y mujeres del data set completo 
#media
from statistics import mean

mean_male_height_cm = mean(df['Male Height in Cm'])
mean_male_height_ft = mean(df['Male Height in Ft'])
mean_female_height_cm = mean(df['Female Height in Cm'])
mean_female_height_ft = mean(df['Female Height in Ft'])
print('mean_male_height_cm =',mean_male_height_cm)
print('mean_male_height_ft =',mean_male_height_ft)
print('mean_female_height_cm =',mean_female_height_cm)
print('mean_female_height_ft =',mean_female_height_ft)

#min
min_male_height_cm = df['Male Height in Cm'].min()
min_male_height_ft = df['Male Height in Ft'].min()
min_female_height_cm = df['Female Height in Cm'].min()
min_female_height_ft = df['Female Height in Ft'].min()
print('min_male_height_cm =',min_male_height_cm)
print('min_male_height_ft =',min_male_height_ft)
print('min_female_height_cm = ',min_female_height_cm)
print('min_female_height_ft =',min_female_height_ft)

#max
max_male_height_cm = df['Male Height in Cm'].max()
max_male_height_ft = df['Male Height in Ft'].max()
max_female_height_cm = df['Female Height in Cm'].max()
max_female_height_ft = df['Female Height in Ft'].max()
print('max_male_height_cm =',max_male_height_cm)
print('max_male_height_ft =',max_male_height_ft)
print('max_female_height_cm =',max_female_height_cm)
print('max_female_height_ft =',max_female_height_ft)
```

```python
stats = df[[
    "Male Height in Cm",
    "Female Height in Cm",
    "Male Height in Ft",
    "Female Height in Ft"
]].agg(["mean", "min", "max"])

print(stats)
```
