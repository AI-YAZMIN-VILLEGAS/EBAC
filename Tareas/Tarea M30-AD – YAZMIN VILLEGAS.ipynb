```python
#pasar a numéricas las columnas que contengan números
df['Joined'] = pd.to_numeric(df['Joined'])
df['Age'] = pd.to_numeric(df['Age'])
df['Overall'] = pd.to_numeric(df['Overall'])
df['Potential'] = pd.to_numeric(df['Potential'])
df['Value'] = pd.to_numeric(df['Value'])
df['Wage'] = pd.to_numeric(df['Wage'])
df['International Reputation'] = pd.to_numeric(df['International Reputation'])
df['Skill Moves'] = pd.to_numeric(df['Skill Moves'])
df['Height'] = pd.to_numeric(df['Height'])
df['Weight'] = pd.to_numeric(df['Weight'])
df['Weight'] = pd.to_numeric(df['Weight'])
df.dtypes
```

Añadir una columna “Years Playing”, que calcule el año actual menos la columna “Joined”

```python
from datetime import datetime

current_year = datetime.now().year


df['Years Playing'] = current_year - df['Joined']
```

Ordenar y mostrar los datos por la columna ReleaseClause (sueldo)

```python
df['Release Clause'] = pd.to_numeric(df['Release Clause'])
df = df.sort_values(by='Release Clause',ascending=True)
df.head()
```

Generar un nuevo dataset que contenga el año (joined) y el número de jugadores (groupby)

```python
df_numero_de_jugadores = df.groupby('Joined').size().reset_index(name='Número de jugadores')
df_numero_de_jugadores.head()
```
