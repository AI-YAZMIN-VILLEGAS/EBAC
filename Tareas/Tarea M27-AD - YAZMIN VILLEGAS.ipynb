Aplicar los 3 puntos anteriores a un ejemplo donde se busca un archivo “test.txt” que NO existe en el directorio donde se está buscando.  
Generar las excepciones y mensajes de error correspondientes. 
Excepciones y las funciones TRY/EXCEPT 
Ejemplos con TRY/EXCEPT/ELSE/FINALLY 

```python
f = open("test.txt”")
```

```python
try:
    f = open("test.txt”")
except FileNotFoundError as err:
    print(err)
```

```python
try:
    f = open("test.txt”")
except FileNotFoundError:
    print("El archivo test.txt” no existe")
except PermissionError:
    print("No tienes permisos para abrir el archivo")
```

```python
try:
    f = open("test.txt”")
except FileNotFoundError as err:
    print(err)
else:
    print("el archivo test.txt” existe")
finally:
    print("el fin")
```

```python
try:
    f = open("test.txt")
except FileNotFoundError as err:
    print("Archivo no encontrado:", err)
else:
    print("Archivo abierto correctamente")
    f.close()
finally:
    print("Intentamos abrir el archivo test.txt")
```

```python
try:
    with open("test.txt") as f:
        contenido = f.read()
except FileNotFoundError:
    print("El archivo test.txt no existe")
else:
    print("Archivo leído correctamente")
finally:
    print("Proceso finalizado")
```
