```python
import csv
import os
from pathlib import Path

data_csv = []

file_name_csv = "fifa_eda.csv"
file_path_csv = Path('.',file_name_csv)
with open(file_path_csv, newline='', encoding='utf-8') as file:
    csv_reader = csv.reader(file)
    for line in csv_reader:
        data_csv.append(line)

first_10_rows_csv = data_csv[:10]
print(first_10_rows_csv)

last_10_rows_csv = data_csv[-10:]
print(last_10_rows_csv)

file_name_csv_writer = "4rowscsv.csv"
path_file_path_csv_writer = Path('.',file_name_csv_writer)
with open(path_file_path_csv_writer, "w", newline='',encoding="utf-8") as file:
    csv_writer = csv.writer(file)
    csv_writer.writerows(data_csv[:5])

data_csv_reader = []
with open(file_name_csv_writer, newline = '', encoding='utf-8') as file:
    csv_reader = csv.reader(file)
    for line in csv_reader:
        data_csv_reader.append(line)
print(data_csv_reader[:5])
```

```python
import json
from pathlib import Path

# Carpeta donde están los JSON
folder_path = Path('./GOT Subtitles')

# Obtener todos los archivos JSON
json_files = list(folder_path.glob('*.json'))
print(f'Total archivos encontrados: {len(json_files)}')


# Procesar cada archivo
for file in json_files:
    print(f'\nProcesando: {file.name}')
    
    # Leer el archivo y guardarlo en una lista
    with open(file, 'r', encoding='utf-8') as f:
        data = json.load(f)
    
    # Asegurarse de que sea lista
    if not isinstance(data, list):
        print('El archivo no contiene una lista de registros.')
        continue
    
    # Obtener las 10 primeras filas
    print('\nPrimeras 10 filas:')
    print(data[:10])
    
    # Obtener las 10 últimas filas
    print('\nÚltimas 10 filas:')
    print(data[-10:])
    
    # Seleccionar 2-4 columnas (ajustar claves según tu dataset)
    if len(data) > 0 and isinstance(data[0], dict):
        keys = list(data[0].keys())[:3]  # tomar 3 columnas
        filtered_data = [{k: item.get(k) for k in keys} for item in data]
        
        # Guardar nuevo archivo
        new_file = file.parent / f'filtered_{file.name}'
        with open(new_file, 'w', encoding='utf-8') as out:
            json.dump(filtered_data, out, indent=4, ensure_ascii=False)
        
        print(f'Archivo generado: {new_file.name}')
    else:
        print('No se pudieron seleccionar columnas (estructura inesperada).')




with open("./GOT Subtitles/"+'result1.json','r', encoding='utf-8') as json_file:
       data = json.load(json_file)
print(data[:10])
```

```python

from pathlib import Path
import openpyxl

file_xlsx = 'Amazon_top100_bestselling_books_2009to2021.xlsx'
file_xlsx_path = Path('.', file_xlsx)

wb_obj = openpyxl.load_workbook(file_xlsx_path)
sheet = wb_obj.active

data_excel = []

for row in sheet.iter_rows(values_only=True):
        data_excel.append(list(row))
    
print("Total filas:", len(data_excel))
    
# Obtener las 10 primeras filas
first_10_rows_xlsx = data_excel[:10]
last_10_rows_xlsx = data_excel[-10:]
print(first_10_rows_xlsx)
print(last_10_rows_xlsx)

```
